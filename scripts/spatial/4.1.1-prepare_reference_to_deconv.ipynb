## Load data

```r
df_info <- read_csv(here::here("utils/info_samples.csv"),show_col_types = FALSE)
```

```r
df_info <- df_info <- df_info %>% 
  dplyr::filter(quality != "Bad") 

set1 <- df_info[df_info$modality == "st", ]$patient
set2 <- df_info[df_info$modality == "multiome", ]$patient

paired_list <- Reduce(intersect, list(set1, set2))
```

```r
for (pat in paired_list) {
  # get shared info
  m_subproj <- df_info[df_info$modality == "multiome" & df_info$patient == pat, ]$subproject
  s_subproj <- df_info[df_info$modality == "st" & df_info$patient == pat, ]$subproject
  
  # read multiome object
  source(here::here(glue::glue("{m_subproj}/multiome_analysis/misc/paths.R")))
  ref_obj <- readRDS(here::here(glue::glue("{m_subproj}/multiome_analysis/3-clustering/{ver}/R_objects_{ver}/level_2/3.3.0-l2_anno_{pat}.rds")))
  
  # clean from unecessary info
  ref_obj[['ATAC']] <- NULL
  
  # write object to st folder
  source(here::here(glue::glue("{s_subproj}/st_analysis/misc/paths.R")))
  "{ann}/{robj_dir}" %>%
  glue::glue() %>%
  here::here() %>%
  dir.create(path = .,
           showWarnings = FALSE,
           recursive = TRUE)
  saveRDS(ref_obj, here::here(glue::glue("{ann}/{robj_dir}/3.3.0-anno_{pat}.rds")))
}
```
