## Convert

```python
for pat in df_info['patient'].tolist():
    for obj in objects:
        if obj == pat+'.h5ad':
            print(pat)
            adata = sc.read_h5ad(os.path.join(path_to_data, pat+".h5ad"))
            
            counts_df = pd.DataFrame(
                adata.layers['counts'].toarray(),
                index=adata.obs_names,
                columns=adata.var_names
            )
            counts_df = counts_df.astype(int)
            
            # Extract metadata
            metadata_df = adata.obs
            
            # Save to CSV
            counts_df.to_csv(os.path.join(path_to_data,"counts_matrix_"+pat+".csv"))
            metadata_df.to_csv(os.path.join(path_to_data,"metadata_"+pat+".csv"))
```
